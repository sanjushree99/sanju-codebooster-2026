```python
import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print('All libraries imported successfully!')
print(f'pandas : {pd.__version__}')
print(f'requests : {requests.__version__}')
```

```python
raw_df = pd.read_csv('messy_sales_data.csv')

print(f'Raw data loaded: {raw_df.shape[0]} rows, {raw_df.shape[1]} columns')
print(f'Columns: {raw_df.columns.tolist()}')
print('\nFirst 5 rows:')
raw_df.head()
```

```python
## Step 3 - Diagnaose ALL data quality problems before fixing ##
print('=' *55)
print(' DATA QUALITY DIAGNOSIS REPORT ')
print('=' *55)

# 1. MISSING VALUES
print('\n1. MISSING VALUES per column')
print(raw_df.isnull().sum())

# 2. Duplicates
print('\n2. DUPLICATES')
print(f'Number of duplicated rows: {raw_df.duplicated().sum()}')

# 3. Data types
print('\n3. DATA TYPES')
print(raw_df.dtypes)

# 4. Unique values in text columns (spot inconsistences)
print('\n[4] UNIQUE CATEGORIES:', raw_df['category'].unique())
print('[4] Sample customer name:', raw_df['customer_name'].dropna().unique()[:8])
print('[4] Sample order_date values:', raw_df['order_date'].unique()[:6])
```

```python
# Missing values for quantity column
print('\n[1] MISSING VALUES in quantity column:')
print(raw_df['quantity'].isnull().sum())
```

```python
# Step 4 - Create a working copy (ETL best Practice)

df = raw_df.copy()
print(f'working Copy created: {df.shape}')
print('raw_df is untouched - we can always reset by running df = raw_df.copyt()')
```

```python
#step 5 - Fix #1: Handle Missing values

print('Before fixing nulls:',df.isnull().sum().sum(),'total missing values')

df['customer_name'].fillna('Unknown', inplace=True)

# Calculate median_qty before filling missing values if you want to use it
median_qty = df['quantity'].median()
df['quantity'].fillna(median_qty, inplace=True)
print(f' Filled missing quantity with median: {median_qty:.2f}')

df['category'].fillna('Unknown', inplace=True)

print('After fixing nulls:',df.isnull().sum().sum(),'total missing values')
```

```python
#Step 6 - Fix #2: Remove Dup;icates

print(f'Before deduplication: {len(df)} rows')
print(f' Duplicate rows: {df.duplicated().sum()}')
print('\nDuplicate rows:')
print(df[df.duplicated(keep=False)][['order_id','customer_name','product','order_date']])

df.drop_duplicates(inplace=True)

print(f']\nAfter deduplication: {len(df)} rows')
print(f'Rows removed: {len(raw_df) - len(df)}')
```

### Finding Duplicates in `student_performanceduple.csv.xlsx`

```python
try:
    # Load the Excel file into a pandas DataFrame
    student_df = pd.read_excel('student_performanceduple.csv.xlsx')
    print(f'Successfully loaded student_performanceduple.csv.xlsx with {student_df.shape[0]} rows and {student_df.shape[1]} columns.')

    # Find duplicate rows
    duplicate_students = student_df[student_df.duplicated(keep=False)]

    if not duplicate_students.empty:
        print(f'\nFound {len(duplicate_students)} duplicate rows in the student data. Here are the duplicate rows:')
        display(duplicate_students)
        print(f'Total number of unique duplicate rows: {len(duplicate_students.drop_duplicates())}')
    else:
        print('\nNo duplicate rows found in the student data.')

except FileNotFoundError:
    print("Error: 'student_performanceduple.csv.xlsx' not found. Please make sure the file is in the correct directory.")
except Exception as e:
    print(f"An error occurred while processing the Excel file: {e}")
```

```python
print('Sample dates before parsing:')
print(df['order_date'].head(8).tolist())

# Convert to datetime object first (keeping dayfirst=True)
df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True,
    format='mixed',
    errors='coerce'
)

nat_count = df['order_date'].isnull().sum()
print(f'\nUnparseable dates (NaT): {nat_count}')

# Extract year, month, and month name
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.strftime('%B')

# Format the order_date column specifically to Day-Month-Year for display
df['order_date'] = df['order_date'].dt.strftime('%d-%m-%Y')

print('\nSample dates after parsing (Day-Month-Year):')
print(df[['order_date', 'year', 'month', 'month_name']].head(5))
```
